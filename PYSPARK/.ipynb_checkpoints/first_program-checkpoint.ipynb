```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("pyspark").master("local[*]").getOrCreate()

data = [
    ("Alice", "Engineering", 85000, 30),
    ("Bob", "Engineering", 92000, 35),
    ("Charlie", "Sales", 60000, 28),
    ("Diana", "Sales", 65000, 32),
    ("Ethan", "HR", 55000, 40),
    ("Fiona", "HR", 58000, 25),
]

columns = ["Name", "Department", "Salary", "Age"]

df = spark.createDataFrame(data, schema=columns)

df.show()
```

```python
df1 = df.select("Name","Department").show()
```

```python
from pyspark.sql.functions import col, when

df1 = df.select(
    when(col("age") > 30,"yes").otherwise("not").alias("is_over_30")
)
df1.show()
```

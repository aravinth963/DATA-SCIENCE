```python
from pyspark.sql import SparkSession
from pyspark.sql import DataFrame
from pyspark.sql.functions import col, when

spark = SparkSession.builder.appName("TRY CATCH BLOCK").master("local[*]").getOrCreate()

def iseligible(df1:DataFrame) -> DataFrame:
    
    return df1.select(
        col("id"),
        col("name"),
        col("age"),
        when(col("age") > 18,"eligible").when((col("age") >= 16) & (col("age") <= 18),"Getting eligible").otherwise("not eligible").alias("status")
    )
    
try:
    mylist = [
      (1, "Balaji", 16),
      (2, "Bala",  66),
      (3, "Sanjay", 76),
      (4, "Karthik" ,17)
    ]

    df = spark.createDataFrame(mylist)
    df1 = df.toDF("id","name","age")

    result = iseligible(df1)
    result.show()
except Exception as err:
    print(f"error occur{err}")
finally:
    print("successfully runned")
    
```
